```python
#Personal Savings Tracker

def update_savings(current_total, deposit):
    result = current_total + deposit
    return result

def check_goal(total, target):
    if total >= target:
        return "Goal reached!"
    else:
        return "Keep saving!"

target = float(input("Enter target for savings: "))
total = 0

deposit = float(input("Enter deposit (0 to finish): "))

while deposit != 0:
      total = update_savings(total, deposit)

      print(f"Updated savings: ₱{total:.2f}")

      deposit = float(input("Enter deposit (0 to finish): "))

print(f"\nTotal savings: ₱{total:.2f}")

if total < target:
    remaining = target - total
    print(f"Remaining amount needed: ₱{remaining:.2f}")

    print(check_goal(total, target))
```

```python
#Classroom Energy Usage Analyzer

def calculate_energy(power_watts, hours_used):
    result = (power_watts * hours_used)/1000
    return result

def classify_usage(energy):
    if energy < 0.5:
       return "Low"
    elif energy <= 2.0:
       return "Moderate"
    else:
       return "High"

number_devices = int(input("How many devices?"))

total_energy = 0
highest_energy = -1
highest_device = ""

for i in range(1, number_devices + 1):

    device_name = input(f"Device {i} name: ")
    power = float(input("Power rating (watts): "))
    hours = float(input("Hours used: "))

energy = calculate_energy(power, hours)
classification = classify_usage(energy)

print(f"Energy consumption: {energy:.2f} kWh")
print(f"Classification: {classification}")

total_energy = total_energy + energy

if energy > highest_energy:
        highest_energy = energy
        highest_device = device_name


print(f"Total energy consumption: {total_energy:.2f} kWh")
print(f"Highest-consuming device: {highest_device}")
```

```python
#Water Refilling Station Sales Tracker

def calculate_sale(quantity, price_per_container):
    result = quantity * price_per_container
    return result

def calculate_discount(transaction):
    if transaction >= 500:
       return transaction * 0.05
    else:
       return 0

total_sales = 0
total_discounts = 0

price_per_container = float(input("Price per container: "))
quantity = int(input("Enter quantity sold (0 to finish): "))

while quantity != 0:

  sale = calculate_sale(quantity, price_per_container)
  discount = calculate_discount(sale)
  total_amount = sale - discount

  print(f"Sale: ₱{sale:.2f}")
  print(f"Discount: ₱{discount:.2f}")
  print(f"Total Amount: ₱{total_amount:.2f}")

  total_sales = total_sales + sale
  total_discounts = total_discounts + discount

  quantity = int(input("Enter quantity sold (0 to finish): "))

  net_sales = total_sales - total_discounts

print("SUMMARY")
print(f"Total sales: ₱{total_sales:.2f}")
print(f"Total discounts: ₱{total_discounts:.2f}")
print(f"Net sales: ₱{net_sales:.2f}")
```

```python
#Reading Time and Progress Analyzer

def calculate_reading_speed(pages, minutes):
    if minutes == 0:
        return 0
    else:
        return pages / minutes

def classify_speed(speed):
    if speed < 1:
      return "Low"
    elif speed <= 2:
      return "Steady"
    else:
      return "Fast"

sessions = int(input("How many reading sessions to be recorded?"))

total_pages_read = 0
total_reading_time = 0
overall_reading_speed = 0

for i in range(1, sessions + 1):
  pages = int(input("Pages: "))
  time = int(input("Time: "))

  speed = calculate_reading_speed(pages, time)
  classification = classify_speed(speed)

  print("Session", i, "Pages Read:", pages)
  print("Minutes Spent:", time)
  print("Reading speed:", pages / time)
  print("Classification:", classification)

  total_pages_read =  total_pages_read + pages
  total_reading_time = total_reading_time + time

if total_reading_time == 0:
    overall_reading_speed = 0
else:
    overall_reading_speed = total_pages_read / total_reading_time

print("SUMMARY")
print("Total Pages Read:", total_pages_read)
print("Total Reading Time:", total_reading_time)
print("Overall Reading Speed:", overall_reading_speed)
```
