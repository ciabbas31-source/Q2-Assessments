```python
# Student Score Analyzer

scores = [85, 90, 78, 92, 88]


def display_scores(scores):
    print("\n=== Student Scores ===")
    for index, score in enumerate(scores):
        print(f"Index {index}: {score}")


def search_score(scores, target):
    found = False

    for index, score in enumerate(scores):
        if score == target:
            print(f"Score {target} found at index {index}.")
            found = True

    if not found:
        print(f"Score {target} was not found.")


def compute_average(scores):
    average = sum(scores) / len(scores)
    print(f"\nAverage Score: {average:.2f}")


def update_score(scores, index, new_score):
    if 0 <= index < len(scores):
        scores[index] = new_score
        print("\nScore successfully updated!")
    else:
        print("\nInvalid index. Please try again.")


def main():
    while True:
        print("\n=== Student Score Analyzer ===")
        print("Select the number of the operation you want to perform:")
        print("1. Display Scores")
        print("2. Search Score")
        print("3. Compute Average Score")
        print("4. Update Score")
        print("5. Display Updated Scores")
        print("6. Exit")

        choice = input("\nEnter your choice: ")

        if choice == "1":
            display_scores(scores)

        elif choice == "2":
            try:
                target = int(input("Enter the score to search: "))
                search_score(scores, target)
            except ValueError:
                print("Please enter a valid number.")

        elif choice == "3":
            compute_average(scores)

        elif choice == "4":
            try:
                index = int(input("Enter the index of the score to update: "))
                new_score = int(input("Enter the new score: "))

                if 0 <= new_score <= 100:
                    update_score(scores, index, new_score)
                else:
                    print("Score must be between 0 and 100.")

            except ValueError:
                print("Please enter valid numbers.")

        elif choice == "5":
            display_scores(scores)

        elif choice == "6":
            print("\nThank you for using the Student Score Analyzer!")
            break

        else:
            print("\nInvalid choice. Please select from 1 to 6.")


main()
```
